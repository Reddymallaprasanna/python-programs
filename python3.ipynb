```python
#sets
set_1={1,2,3,4,5,6}
print(set_1)
print("add and remove elements")
set_1.add(77)
print(set_1)
set_1.remove(5)
print(set_1)
print(6 in set_1)
print(1 not in set_1)
```

```python
#set operationss
a={1,2,3,4}
b={4,5,6,7}
print("union :",a.union(b))
print("intersection:", a.intersection(b))
print("differnece:",a.difference(b))
print("symmentric difference:", a.symmetric_difference(b))
```

```python
##subset and super set 
a={1,2,3}
b={1,2,3,4,5,6}
print("a is sub set to b:",a.issubset(b))
print("b is sub set of a:",b.issubset(a))
```

```python
#set operations using conditional operators
a={1,2,3}
b={1,2,3,4,5}
print("sub set or not:",a<=b)
print("proper superset:",a<b)
print("superset or not:",a>=b)
print("proper superset:",b>a)
```
